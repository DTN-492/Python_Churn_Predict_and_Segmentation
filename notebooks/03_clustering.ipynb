# 0. Load

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.impute import KNNImputer
import os
import json
```

```python
df = pd.read_csv("../Data/churn_processed.csv")
df = df[df['Churn'] == 1]
df = df.drop(columns = ["CustomerID", "Churn"])
df_churn = df.copy()
```

# 1. Preprocessing

## a. Encode

```python
df_churn['CouponUsed'] = df_churn['CouponUsed'].fillna(0) # CouponUsed missing => Customer hasn't use any coupon
df_churn['OrderCount'] = df_churn['OrderCount'].fillna(0) # Customer hasn't order any items
df_churn['DaySinceLastOrder'] = df_churn['DaySinceLastOrder'].fillna(9999) # Customer hasn't order any items => fill with big number to separate them with other customers
```

```python
# Impute other columns with KNNimputer
df_churn['CityTier'] = df_churn['CityTier'].astype('str')
df_encode = pd.get_dummies(df_churn, drop_first = True)
df_encode = pd.DataFrame(KNNImputer(n_neighbors = 2).fit_transform(df_encode), columns = df_encode.columns, index = df_encode.index)
df_encode.head()
```

## b. Scale

```python
# Scale numeric columns using Z-score scaling (x-μ/σ)
col = df_encode.select_dtypes(include = 'number').columns[1:]
df_encode_scale = df_encode.copy()

for c in col:
    m = df_encode[c].mean()
    sd = df_encode[c].std()
    df_encode_scale[c] = (df_encode[c] - m)/sd
df_encode_scale.head()
```

## c. PCA

```python
explain_variance = []
for i in range (1, 15):
  pca = PCA(n_components = i)
  pca.fit(df_encode_scale)
  explain_variance.append(sum(pca.explained_variance_ratio_))
plt.style.use('dark_background')
sns.lineplot(x = list(range(1,15)), y = explain_variance, color = "#12aeff")
plt.grid(color = "#eeeeee")
```

Ideal number of components here is 9 (explain ~ 80% of variance)

```python
# Get first 9 components/features
pca = PCA(n_components = 9)
pca.fit(df_encode_scale)
PCA_ds = pd.DataFrame(pca.transform(df_encode_scale), columns = (["col1","col2", "col3", "col4", "col5","col6", "col7", "col8", "col9"]))
PCA_ds.head()
```

## c1. PCA scale

```python
# Z-score scaling with PCA data
PCA_scale = PCA_ds.copy()

for c in PCA_ds.columns:
    m = PCA_ds[c].mean()
    sd = PCA_ds[c].std()
    PCA_scale[c] = (PCA_ds[c] - m)/sd
PCA_scale.head()
```

## c2. PCA norm

```python
# Min-max scaling with PCA data
PCA_norm = PCA_ds.copy()

for c in PCA_ds.columns:
    mn = PCA_ds[c].min()
    mx = PCA_ds[c].max()
    PCA_norm[c] = (PCA_ds[c] - mn)/(mx - mn)
PCA_norm.head()
```

# 2. K-means

```python
ss = []
max_clusters = 10
for i in range(1, max_clusters + 1):
    kmeans = KMeans(n_clusters = i, init = 'k-means++', random_state = 10)
    kmeans.fit(PCA_scale)
    ss.append(kmeans.inertia_)

# Plot the Elbow method
plt.figure(figsize = (10,5))
plt.plot(range(1, max_clusters + 1), ss, marker = 'o', linestyle = '--')
plt.title('Elbow Method (Z-score scaled data)')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
ss = []
max_clusters = 10
for i in range(1, max_clusters + 1):
    kmeans = KMeans(n_clusters = i, init = 'k-means++', random_state = 10)
    kmeans.fit(PCA_norm)
    ss.append(kmeans.inertia_)

# Plot the Elbow method
plt.figure(figsize = (10,5))
plt.plot(range(1, max_clusters + 1), ss, marker = 'o', linestyle = '--')
plt.title('Elbow Method (Min-max scaled data)')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

There is no optimal value for k, so segmentation churned customers is not necessary
